# 1. Qu’est-ce que la Programmation Fonctionnelle ?

La programmation fonctionnelle est un paradigme où :

- les **fonctions** sont au centre de votre logique.
- les données sont **immuables** ou traitées comme telles,
- on évite les **effets de bord** (on évite de modifier des variables globales)
- on compose des transformations successives plutôt que de modifier des objets.

En Programmation Fonctionnelle, un programme ressemble à une **suite d’applications de fonctions**.

# 2. Programmation Fonctionnelle vs Programmation Orientée Objet

### Programmation Fonctionnelle
- Centrée sur les **fonctions pures**.
- Données traitées sans modification directe.
- Idéale pour les transformations de données, les pipelines, les calculs.

### Programmation Orientée Objet
- Centrée sur les **objets** et leur état.
- Encapsulation et interactions entre instances.
- Idéale pour représenter des systèmes complexes avec comportements partagés.

### Quand utiliser quoi ?

| Besoin | PF recommandée | POO recommandée |
|--------|----------------|-----------------|
| Pipelines de transformations | ✔️ | |
| Modéliser des entités avec état | | ✔️ |
| Code court et déclaratif | ✔️ | |
| Architecture large et structurée | | ✔️ |
| Éviter les effets de bord | ✔️ | |

En pratique :  
→ **POO** pour les systèmes structurés (jeux, API, simulations, ML frameworks).  
→ **PF** pour traiter, filtrer, transformer des données (listes, tableaux, flux).

# Fonctions Lambda

Une **fonction lambda** est une fonction anonyme définie en une seule ligne.

### Syntaxe
```py
lambda arguments: expression
```

### Pourquoi s’en servir ?

- L'intérêt des fonctions lambda en Python est de pouvoir créer des petites fonctions anonymes, concises et jetables à la volée, notamment pour des opérations simples et contextuelles
- Évite de définir une fonction nommée pour des usages ponctuels.
- Utile avec map, filter, reduce ou comme argument de tri.

### Exemple Simple

Calculons la différence au carré entre deux nombres :

````python
# Fonction lambda simple
diff_quadratique = lambda x, y: (x - y) ** 2

# Utilisation
print(diff_quadratique(3, 5)) 
````

````python
# Comparaison avec une fonction classique
def diff_quadratique_classique(x, y):
    return (x - y) ** 2

print(diff_quadratique_classique(3, 5))
````

## Utilisation avec `map`

La fonction **`map()`** applique une fonction à chaque élément d’un itérable.

### Syntaxe
```py
map(fonction, itérable)
```

### Pourquoi s’en servir ?

Transforme une collection sans boucle explicite.
Concis et lisible pour des opérations uniformes.

### Exemple

````python
nombres = [5, 12, 79, 2, 46, 123, 6, 71, 89]
````

````python
list(map(lambda x: x**2, nombres))
````

## Utilisation avec `filter`

La fonction **`filter()`** sélectionne les éléments d’un itérable qui satisfont une condition.

### Syntaxe
```python
filter(fonction, itérable)
```

### Pourquoi s’en servir ?

- Filtre une collection de manière concise.
- Évite les boucles avec conditions redondantes.

### Exemple

Filtrons les sorties d'un modèle de classification pour ne garder que celles au-dessus d'un seuil.

````python
nombres = [0.4, -1.5, -2.6, 1.7, 1.8, 0.6, -0.2]
````

````python
def fonction_filtre(x):
    return x < 0
````

````python
positives = list(filter(fonction_filtre, nombres))
````

````python
positives
````

### Comparaison avec une fonction traditionnelle

````python
# Avec une fonction nommée
def est_positif(x):
    return x > 0
````

````python
# Sans filter (boucle classique)
positives = []
for n in nombres:
    if est_positif(n):
        positives.append(n)
````

````python
print("Valeurs supérieures à 0.2 (boucle) :", positives)
````

## Utilisation avec `reduce`

La fonction **`reduce()`** (du module `functools`) applique une fonction cumulativement à une séquence pour réduire ses éléments à une seule valeur.

L'idée est la suivante : 
```
reduce(func, [a, b, c, d])

# devient

func(func(func(a, b), c), d)
```

### Syntaxe
```python
from functools import reduce
reduce(fonction, itérable, initial)
```

### Pourquoi s’en servir ?

- Réduit une collection à une valeur unique (ex. : somme, produit).
- Puissant pour les opérations cumulatives.

### Exemple

````python
from functools import reduce
````

````python
nums = [1, 2, 3, 4]

product = reduce(lambda x, y: x * y, nums)

print(product)
````

````python
lists = [[1, 2], [3, 4], [5]]

flattened = reduce(lambda x, y: x + y, lists)

print(flattened)
````

````python
cart = [{"price": 19.99}, {"price": 5.50}, {"price": 3.49}]

total_price = reduce(lambda x, y: x + y["price"], cart, 0)

print(round(total_price, 2))
````

## En résumé : Pourquoi Utiliser `lambda`, `map`, `filter`, `reduce` ?

### Avantages
1. **Concision** : Réduit le code par rapport aux boucles explicites.
2. **Lisibilité** : Exprime clairement l’intention (transformation, filtrage, réduction).
3. **Fonctionnalité** : Encourage un style fonctionnel, évitant les effets secondaires.
4. **Performance** : `map` et `filter` sont optimisés pour les itérations (bien que les compréhensions de listes soient souvent aussi rapides).

### Limites
- Moins intuitif pour les débutants.
- Peut être moins flexible pour des logiques complexes.
- `reduce` est moins utilisé depuis l’essor des compréhensions et `sum()`.

### Quand les Utiliser ?
- **Petites transformations rapides** (ex. : `map` pour appliquer une fonction).
- **Filtres simples** (ex. : `filter` pour sélectionner).
- **Réductions spécifiques** (ex. : `reduce` pour des calculs cumulatifs inhabituels).

## Correction

````python
# 1
individual_revenues = list(map(lambda sale: sale['price'] * sale['quantity'], sales_data))

grand_total_revenue = reduce(lambda total, revenue: total + revenue, individual_revenues)

print(f"1. Individual Revenues: {individual_revenues}")
print(f"1. Grand Total Revenue: ${grand_total_revenue:.2f}")
````

````python
# 2
high_value_sales = list(filter(lambda sale: sale['price'] * sale['quantity'] > 300.00, sales_data))

print(f"\n2. High-Value Sales (Total Revenue > $300):")
for sale in high_value_sales:
    print(f"   - Product: {sale['product_id']}, Region: {sale['region']}, Revenue: ${sale['price'] * sale['quantity']:.2f}")
````

````python
# 3
lower_case_ids = list(map(lambda sale: sale['product_id'].lower(), sales_data))

print(f"\n3. Lowercase Product IDs: {lower_case_ids}")
````

````python
# 4
north_sales = filter(lambda sale: sale['region'] == 'North', sales_data)

north_quantities = map(lambda sale: sale['quantity'], north_sales)

max_north_quantity = reduce(lambda a, b: a if a > b else b, north_quantities)

print(f"\n4. Maximum Quantity Sold in the 'North' Region: {max_north_quantity}")
````
