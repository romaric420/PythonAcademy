# Concepts de Programmation Orientée Objet (POO) en Python

Dans cette section, nous allons explorer les **concepts fondamentaux de la POO** en Python : classes, objets, attributs, méthodes, méthodes spéciales (`__init__`, `__str__`, `__repr__`), encapsulation, héritage et polymorphisme. La POO est essentielle pour structurer des programmes complexes de manière claire et réutilisable.

## Qu’est-ce que la POO ?
La **Programmation Orientée Objet** est un paradigme qui organise le code autour d’**objets**, instances de **classes**, combinant données (**attributs**) et comportements (**méthodes**). Python est un langage qui supporte pleinement la POO.


## Exemple :

**Classe** : voiture
- *attributs*
    - couleur
    - modele
    - année
    - immatriculation
- *méthodes*
    - avancer
    - reculer
    - freiner
    - tourner a gauche
    - tourner a droite

## Pourquoi Utiliser la Programmation Orientée Objet (POO) ?

- **Modélisation complexe** : Représenter des entités réelles (ex. : utilisateurs, véhicules) avec des propriétés et comportements.
- **Réutilisabilité** : Héritage pour partager du code entre classes similaires.
- **Encapsulation** : Protéger les données et contrôler leur accès.

### Par Rapport à d’Autres Approches
- **Vs Fonctions seules** : Les fonctions sont simples mais ne regroupent pas données et comportements ; la POO offre une organisation plus naturelle pour des systèmes complexes.
- **Vs Structures simples** : Les dictionnaires ou tuples peuvent stocker des données, mais sans méthodes associées, contrairement aux objets POO.
- **Vs Programmation fonctionnelle** : La POO privilégie l’état et la mutation, tandis que la fonctionnelle évite les effets secondaires ; choisissez selon la nature du projet.

# Autres Exemples d'utilisation
La POO est parfaitement adaptée pour créer des programmes des genres suivants :

- Un jeu vidéo, avec des personnages (objets) dont les attributs pourraient etre les points de vies, les points d'attaque, et la vitesse de déplacement, et les méthodes seraient par exemple "attaquer", "défendre", "bouger".

- En Deep Learning, quand on développera des réseaux de neurones artificiels, on preferera en général définir une classe qui integre toutes les spécificités sous formes d'attributs (nombre de couches, parametres etc.) et les méthodes (entrainer, évaluer, etc.). De cette maniere

- (Pour finir, une sorte de contre-exemple) : Un gestionnaire de parking, avec des véhicules de tout type (voiture, motos), dont les attributs seraient le modele, l'immatriculation, la couleur, l'état de la voiture, et le numéro de place attribué, et les méthodes pourraient etre : "déplacer", "afficher les informations", ou encore "supprimer" (afin de rendre la voiture a son propriétaire) -> Pourquoi est-ce en quelque sorte un contre-exemple ?


# Quand utiliser / ne PAS utiliser la POO ?

#### La POO est utile quand: 
- le domaine est naturellement fait d’objets
- on a beaucoup de comportement partagé
- l’encapsulation nous simplifie vraiment la vie

#### la POO est a éviter quand :
- On effectue des calculs purs, transformations de données
- pipelines de traitement (données → données → données)
- scripts courts

#### Une règle simple qui marche bien :

Si vous vous retrouvez à créer des classes juste “parce qu’il faut des classes”, c’est le signe que la POO n’est peut-être pas adaptée ici.
En revanche, si le fait de créer une classe clarifie vraiment votre raisonnement… alors c'est que vous etes sur la bonne voie.

## Classes, Objets, Attributs et Méthodes

### Définitions
- **Classe** : Un modèle ou blueprint définissant les attributs (données) et méthodes (comportements).
- **Objet** : Une instance spécifique d’une classe.
- **Attributs** : Variables associées à un objet (ex. : nom, âge).
- **Méthodes** : Fonctions définies dans une classe, opérant sur ses attributs.

### Exemple Simple
Créons une classe `Personne` pour représenter une personne.
- *attributs*:
    - prénom
    - age
- *méthodes*:
    - se_présenter

````python
class Personne:
    
    def __init__(self, nom: str, age: int, city="inconnue"): # Constructeur
        self.nom = nom
        self.age = age
        self.ville = city

        self.se_presenter()

    def se_presenter(self):
        print(f"Je m'appelle {self.nom}, et j'ai {self.age} ans.")
````

````python
alice = Personne(nom="Alice", age=32, city="Paris")
pierre = Personne(nom="Pierre", age=45)
````

````python
alice.se_presenter()
````

````python
pierre.se_presenter()
````

## En résumé:

- **Classe `Personne`** : Définit le modèle avec `__init__` pour initialiser les attributs.
- **Objets** : `alice` et `bob` sont des instances distinctes avec leurs propres valeurs.
- **`self`** : Référence à l’instance courante, obligatoire dans les méthodes.
- **Attributs** : `nom` et `age` sont stockés dans chaque objet via `self`.
- **Méthode `se_presenter`** : Utilise les attributs pour générer une sortie.
- `__init__` : Constructeur, initialise l’objet.



Passons aux méthodes spéciales !

## Méthodes Spéciales

Les **méthodes spéciales** commencent et finissent par `__`. Elles permettent de personnaliser le comportement des objets pour des opérations comme l’initialisation, la représentation ou les comparaisons.

### Méthodes Courantes
- `__init__` : Constructeur, initialise l’objet.
- `__str__` : Représentation lisible (pour `print`).
- `__repr__` : Représentation technique (pour débogage).

### Exemple
Ajoutons ces méthodes à `Personne`.

````python
class Personne:
    """Classe représentant une personne avec un nom et un âge."""
    
    def __init__(self, nom: str, age: int):
        """Initialise une personne avec un nom et un âge.
        
        Args:
            nom (str): Nom de la personne.
            age (int): Âge de la personne.
        """
        self.nom = nom
        self.age = age
    
    def __str__(self) -> str:
        """Retourne une représentation lisible de la personne."""
        return f"{self.nom}, {self.age} ans"
    
    def __repr__(self) -> str:
        """Retourne une représentation technique de la personne."""
        return f"Personne(nom='{self.nom}', age={self.age})"
````

````python
# Tests
alice = Personne("Alice", 25)
````

````python
print(alice) # Utilise __str__ 
````

````python
print(repr(alice)) # Utilise __repr__ 
````

````python
alice.age
````

## Encapsulation

L’**encapsulation** consiste à protéger les données d’un objet en contrôlant leur accès. En Python, elle est conventionnelle (pas strictement privée comme en Java).

### Conventions
- **Public** : `nom` (accessible partout).
- **Protégé** : `_nom` (convention pour indiquer "usage interne").
- **Privé** : `__nom` (name mangling pour éviter les accès directs).

### Exemple
Protégeons le salaire d’un employé.

````python
class Employee:
    
    def __init__(self, nom: str, departement: str, salaire: float):
        self.nom = nom # Public
        self.departement = departement # Public
        self._pourcentageprime = 0.05 # Protégé
        self.__salaire = salaire # Privé

    def calculate_prime(self):
        return self._pourcentageprime * self.__salaire
````

````python
clara = Employee("Clara", "info", 4000.0)
````

````python
clara._Employee__salaire
````

### Méthodes `get` et `set`
Dans la plupart des langages de programmation, on utilise conventionnellement les méthodes `get` et `set` pour "obtenier" (`get`) et "modifier" (`set`) les attributs d'un objet.
Par exemple, dans *Scikit-learn*, l'API propose une méthode `set_params` pour tous ses modeles de machine learning, afin de guider les utilisateurs

````python
class Employee:
    
    def __init__(self, nom: str, departement: str, salaire: float):
        self.name = nom # Public
        self.departement = departement # Public
        self._pourcentageprime = 0.05 # Protégé
        self.__salaire = salaire # Privé

    def calculate_prime(self):
        return self._pourcentageprime * self.__salaire


    def get_salary(self):
        return self.__salaire

    def set_salary(self, nouveau_salaire):
        self.__salaire = nouveau_salaire
````

````python
clara = Employee("Clara", "info", 4000.0)
````

````python
clara.set_salary(4500)

clara.get_salary()
````

## Héritage

L’**héritage** permet à une classe d’hériter des attributs et méthodes d’une autre classe (classe parent).

#### Mise en situation

On dispose de 2 classes:

**Voiture**
- Attributs:
    - constructeur
    - modele
    - couleur
    - passagers (liste)
- Méthodes:
    - démarrer()
    - arreter()
    - avancer()
    - freiner()
    - monter()
    - descendre()


**Camion**
- Attributs:
    - constructeur
    - modele
    - couleur
    - inventaire (liste)
- Méthodes:
    - démarrer()
    - arreter()
    - avancer()
    - freiner()
    - charger()
    - décharger()

#### La Solution: Créer une classe parent, qui inclut les éléments communs

### Syntaxe
```python
class Enfant(Parent):
    ...
```

### Exemple

Créons une classe `Manager` héritant de `Employe`.

````python
class Employee:
    """Classe de base pour un employé."""
    
    def __init__(self, nom: str, salaire: float, pourcentageprime: float):
        self.nom = nom
        self._pourcentageprime = pourcentageprime
        self.__salaire = salaire
    
    def calculer_prime(self):
        return self._pourcentageprime * self.__salaire
    
````

````python
marc = Employee("Marc", 3000, 0.05)
````

````python
marc.calculer_prime()
````

````python
class Manager(Employee):
    """Classe dérivée représentant un manager."""
    
    def __init__(self, nom: str, salaire: float, pourcentageprime: float):
        super().__init__(nom, salaire, pourcentageprime)  # Appelle le constructeur parent
        self.equipe = []

    def rajouter_membre(self, membre):
        self.equipe.append(membre)

    def enlever_membre(self, membre):
        self.equipe.pop(membre)
````

````python
anne = Manager("Anne", 5000, 0.05)
````

````python
anne.rajouter_membre(marc)
````

````python
anne.equipe[0].nom
````

````python
anne.calculer_prime()
````

## En résumé : 

- **`Employe`** : Classe parent avec des attributs et méthodes de base.
- **`Manager`** : Classe enfant qui hérite via `(Employe)` et ajoute `equipe`.
- **`super()`** : Appelle `__init__` du parent pour initialiser `nom` et `salaire`.

## Polymorphisme

Le **polymorphisme** permet à des objets de différentes classes de répondre différemment à une même méthode, grâce à la redéfinition ou à l’héritage.

### Exemple
Nous allons redéfinir le calcul de la prime pour qu'il soit différent selon le role :

- Pour un employe conventionel : `salaire * pourcentage de prime`
- Pour un commercial: `commissions sur ses ventes`
- Pour un manager : `salaire * (pourcentage de prime * perfomances)`

````python
class Employee:
    """Classe de base pour un employé."""
    
    def __init__(self, nom: str, salaire: float, pourcentageprime: float):
        self.nom = nom
        self._pourcentageprime = pourcentageprime
        self._salaire = salaire
    
    def calculer_prime(self):
        return self._salaire * self._pourcentageprime 
    
````

````python
class Commercial(Employee):

    def __init__(self, nom: str, salaire: float, pourcentageprime: float, ventes: float):
        super().__init__(nom, salaire, pourcentageprime)  # Appelle le constructeur parent
        self.ventes = ventes

    def calculer_prime(self):
        return self.ventes * self._pourcentageprime
````

````python
class Manager(Employee):

    def __init__(self, nom: str, salaire: float, pourcentageprime: float, performance: float):
        super().__init__(nom, salaire, pourcentageprime)  # Appelle le constructeur parent
        self.performance = performance

    def calculer_prime(self):
        return self._salaire * self._pourcentageprime * self.performance
````

````python
Antoine = Employee("Antoine", 3000, 0.05)
Younes = Commercial("Younes", 6000, 0.05, 120_000)
Sophie = Manager("Sophie", 5000, 0.1, 0.8)
````

````python
Antoine.calculer_prime()
````

````python
Younes.calculer_prime()
````

````python
Sophie.calculer_prime()
````

## Pourquoi Python est entierement construit sur des objets ?

Il y a plusieurs raisons, et elles sont toutes assez cohérentes quand on se rappelle de la philosophie de Python

- le langage a été conçu pour que tout soit représenté comme un objet avec un type cohérent. Ça simplifie énormément la logique interne, et fait en sorte que toute entité se manipule de la meme maniere.
- Cela débloque de nombreuses fonctionnalités:
    - utiliser des fonctions comme arguments
    - retourner des fonctions depuis d’autres fonctions
- Du coup, en tant que dev, nous n’avons pas à nous demander “ok, ce truc, est-ce que je peux le mettre dans une liste ? le passer à un paramètre ? l’attacher à un objet ?”. La réponse est toujours : oui.

#### fonction passée en argument

````python
def saluer(nom):
    print(f"Salut {nom} !")

def appliquer_sur_nom(fonction, nom):
    fonction(nom)

appliquer_sur_nom(saluer, "Alice")
````

#### fonction retournée par une autre fonction

````python
def creer_multiplicateur(x):
    def multiplier(y):
        return x * y
    return multiplier

fois_3 = creer_multiplicateur(3)

print(fois_3(10))
````

## Correction

````python
from typing import List


class Vehicule:
    """Classe de base pour un véhicule."""
    
    def __init__(self, marque: str, vitesse_max: float):
        self._marque = marque
        self.__vitesse_max = vitesse_max
    
    def __str__(self) -> str:
        return f"Véhicule {self._marque}"
    
    def __eq__(self, autre: 'Vehicule') -> bool:
        # deux véhicules sont égaux s'ils ont la même marque et vitesse max (tous les attributs)
        return self._marque == autre._marque and self.__vitesse_max == autre.__vitesse_max
    
    def deplacer(self) -> str:
        return f"Le véhicule {self._marque} se déplace."


class Voiture(Vehicule):
    """Classe dérivée pour une voiture."""
    
    def __init__(self, marque: str, vitesse_max: float, portes: int):
        super().__init__(marque, vitesse_max)
        self.portes = portes
    
    def deplacer(self) -> str:
        return f"La voiture {self._marque} roule à {self.portes} portes."


class Moto(Vehicule):
    """Classe dérivée pour une moto."""
    
    def deplacer(self) -> str:
        return f"La moto {self._marque} fonce !"
````

````python
vehicules: List[Vehicule] = [
    Voiture("Peugeot", 180.0, 4),
    Moto("Honda", 200.0)
]
````

````python
for v in vehicules:
    print(v.deplacer())
````

````python
v1 = Voiture("Toyota", 150.0, 4)
v2 = Voiture("Toyota", 150.0, 5)
````

````python
print(v1 == v2)  
````

- **Encapsulation** : `_marque` (protégé), `__vitesse_max` (privé).
- **Héritage** : `Voiture` et `Moto` héritent de `Vehicule`.
- **Polymorphisme** : `deplacer()` varie selon la classe.
- **Méthodes spéciales** :
  - `__str__` : Représentation personnalisée.
  - `__eq__` : Comparaison d’égalité basée sur les attributs.
- **Typage** : Annotations pour les paramètres et retours.
