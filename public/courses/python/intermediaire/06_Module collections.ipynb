# Collections en Python : Counter, defaultdict, OrderedDict, etc.

Dans cette section, nous allons explorer le module **`collections`**, qui fournit des structures de données spécialisées améliorant les types built-in comme les listes, dictionnaires et tuples. Nous nous concentrerons sur **`Counter`**, **`defaultdict`**, **`OrderedDict`**, et d'autres outils utiles.

## Qu’est-ce que `collections` ?
Le module `collections` offre des alternatives optimisées aux structures de données standard, adaptées à des besoins spécifiques comme le comptage, la gestion de valeurs par défaut ou l’ordre des éléments.

Commençons par `Counter` !

## Utilisation de `Counter`

`Counter` est une sous-classe de dictionnaire conçue pour compter les occurrences d’éléments dans une séquence.

### Importation
```py
from collections import Counter
```

### Fonctionnalités

- Compte automatiquement les éléments.
- Supporte des opérations comme l’addition ou la soustraction de compteurs.

### Exemple Simple

Comptons les éléments d’une liste.

````python
from collections import Counter
````

````python
# Compter les éléments d’une liste
fruits = ["pomme", "banane", "pomme", "orange", "banane", "pomme"]
````

````python
compteur = Counter(fruits)
````

````python
print("Compte des fruits :", compteur)  
````

````python
# Accéder à un élément
print("Nombre de pommes :", compteur["pomme"])  
````

````python
# Éléments absents 
print("Nombre de kiwis :", compteur["kiwi"])  
````

````python
# Liste des éléments les plus fréquents
print("Top 2 :", compteur.most_common(2)) 
````

### Opérations avec `Counter`
`Counter` permet des opérations arithmétiques comme l’addition ou l’intersection.

### Exemple
Comptons et combinons deux ensembles.

````python
from collections import Counter
````

````python
# Deux compteurs
c1 = Counter("aabbbcc")
c2 = Counter("abbc")
````

````python
print("c1 :", c1) 
````

````python
print("c2 :", c2) 
````

````python
# Addition
total = c1 + c2
print("Total (addition) :", total)  
````

````python
# Intersection (minimum)
commun = c1 & c2
print("Intersection (&) :", commun)  
````

````python
# Différence (soustraction)
difference = c1 - c2
print("Différence (-) :", difference)  
````

## Utilisation de `defaultdict`

`defaultdict` est une sous-classe de dictionnaire qui fournit une valeur par défaut pour les clés absentes, évitant les `KeyError`.

### Importation
```python
from collections import defaultdict
```

### Fonctionnalités

- Définit une usine (factory) pour les valeurs par défaut (ex. : int, list).
- Simplifie l’initialisation de structures complexes.

### Exemple Simple

Regroupons des éléments par catégorie.

````python
from collections import defaultdict
````

````python
# Dictionnaire avec liste par défaut
groupes = defaultdict(list)
````

````python
# Ajouter des éléments sans initialisation explicite
donnees = [("fruit", "pomme"), ("légume", "carotte"), ("fruit", "banane")]
for categorie, item in donnees:
    groupes[categorie].append(item)
````

````python
print("Groupes :", dict(groupes))  
````

````python
# Comparaison avec dict classique
d = {}
d["inconnu"].append("test")
````

````python
# Avec defaultdict, pas d’erreur
groupes["inconnu"].append("test")
print("Avec inconnu :", dict(groupes))
````

### `defaultdict` avec `int`
Utilisons `int` comme usine pour compter (similaire à `Counter`).

### Exemple
Comptons des occurrences.

````python
from collections import defaultdict
````

````python
# Compteur avec defaultdict(int)
compteur = defaultdict(int)

texte = "abracadabra"
for lettre in texte:
    compteur[lettre] += 1

print("Compte des lettres :", dict(compteur))  
````

## Utilisation de `OrderedDict`

`OrderedDict` est une sous-classe de dictionnaire qui conserve l’ordre d’insertion des clés (utile avant Python 3.7, où les dictionnaires standards ne le faisaient pas).

### Importation
```python
from collections import OrderedDict
```

### Fonctionnalités

- Maintient l’ordre des éléments.
- Utile pour des structures ordonnées spécifiques.

### Exemple Simple

Créons un dictionnaire ordonné.

````python
from collections import OrderedDict
````

````python
# OrderedDict
od = OrderedDict()
od["c"] = 3
od["a"] = 1
od["b"] = 2
````

````python
print("OrderedDict :", od)  
````

````python
# Itération dans l’ordre d’insertion
for cle, valeur in od.items():
    print(f"{cle} : {valeur}")
````

````python
# Comparaison avec dict standard (depuis 3.7, ordre conservé aussi)
d = {"c": 3, "a": 1, "b": 2}
print("Dict standard :", d)  
````

## Autres Outils de `collections`

### `namedtuple`
Crée des tuples avec des noms d’attributs pour une meilleure lisibilité.

### `deque`
Liste doublement chaînée pour des ajouts/suppressions rapides aux extrémités.

### Exemple avec `namedtuple`
Représentons un point en 2D.

````python
from collections import namedtuple
````

````python
# Définir un namedtuple
Point = namedtuple("Point", ["x", "y"])
````

````python
# Créer une instance
p = Point(3, 4)
print("Point :", p)         
print("X :", p.x, "Y :", p.y) 
````

````python
# Immutabilité
p.x = 5 
````

### Exemple avec `deque`
Utilisons `deque` pour une file à double entrée.

````python
from collections import deque
````

````python

# Créer une deque
d = deque([1, 2, 3])
print("Deque initial :", d)  
````

````python
# Ajouter à gauche et à droite
d.append(4)        # Droite
d.appendleft(0)    # Gauche
print("Après ajouts :", d)  
````

````python
# Supprimer
d.pop()            # Droite
d.popleft()        # Gauche
print("Après suppressions :", d)  
````

## Correction

````python
from collections import Counter
````

````python
# comptage des transactions par catégorie
categories = Counter(transaction[1] for transaction in transactions)
print(f"Transactions par catégorie: {categories}")
````

````python
# identification des 3 produits les plus vendus
produits = Counter(transaction[2] for transaction in transactions)
print(f"Top 3 produits: {produits.most_common(3)}")
````

````python
# répartition des types de clients
types_clients = Counter(transaction[4] for transaction in transactions)
print(f"Répartition clients: {types_clients}")
````

````python
from collections import defaultdict
````

````python
# Regrouper les transactions par date
transactions_par_date = defaultdict(list)
for date, categorie, produit, montant, client_type in transactions:
    transactions_par_date[date].append((categorie, produit, montant, client_type))
````

````python
print("Transactions par date:")
for date, items in transactions_par_date.items():
    print(f"{date}: {items}")
````

````python
# Calculer le chiffre d'affaires total par catégorie
chiffre_affaires = defaultdict(float)
for _, categorie, _, montant, _ in transactions:
    chiffre_affaires[categorie] += montant
````

````python
print("Chiffre d'affaires par catégorie:")
for categorie, total in chiffre_affaires.items():
    print(f"{categorie}: {total:.2f} €")
````

````python
# Créer un mapping client_type → liste des produits achetés
produits_par_client = defaultdict(list)
for _, _, produit, _, client_type in transactions:
    produits_par_client[client_type].append(produit)
````

````python
print("Produits achetés par type de client:")
for client_type, produits in produits_par_client.items():
    print(f"{client_type}: {produits}")
````

````python
from collections import namedtuple
````

````python
# Créer une structure Transaction
Transaction = namedtuple("Transaction", ["date", "categorie", "produit", "montant", "client_type"])
````

````python
# Recréer le dataset avec cette structure
transactions_struct = [
    Transaction("2023-01-15", "Électronique", "Smartphone", 599.99, "Premium"),
    Transaction("2023-01-15", "Vêtements", "T-shirt", 19.99, "Standard"),
    Transaction("2023-01-16", "Électronique", "Laptop", 899.99, "Premium"),
    Transaction("2023-01-16", "Électronique", "Smartphone", 599.99, "Standard"),
    Transaction("2023-01-17", "Alimentation", "Café", 4.50, "Standard"),
    Transaction("2023-01-17", "Vêtements", "Jean", 49.99, "Premium"),
    Transaction("2023-01-18", "Électronique", "Smartphone", 599.99, "Premium"),
    Transaction("2023-01-18", "Alimentation", "Thé", 3.25, "Standard"),
]
````

````python
# Calculer le panier moyen par type de client
panier_moyen = defaultdict(float)
nombre_transactions = defaultdict(int)

for transaction in transactions_struct:
    panier_moyen[transaction.client_type] += transaction.montant
    nombre_transactions[transaction.client_type] += 1

for client_type, total in panier_moyen.items():
    panier_moyen[client_type] = total / nombre_transactions[client_type]

print("Panier moyen par type de client:")
for client_type, montant in panier_moyen.items():
    print(f"{client_type}: {montant:.2f} €")
````

````python
# Identifier le jour avec le plus gros chiffre d'affaires
chiffre_affaires_journalier = defaultdict(float)

for transaction in transactions_struct:
    chiffre_affaires_journalier[transaction.date] += transaction.montant

jour_max = max(chiffre_affaires_journalier, key=chiffre_affaires_journalier.get)
print(f"Jour avec le plus gros chiffre d'affaires: {jour_max} ({chiffre_affaires_journalier[jour_max]:.2f} €)")
````
